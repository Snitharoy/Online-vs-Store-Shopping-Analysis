# <font color="darkblue">Data Preprocessing</font>

### <font color="black">Create Age Group</font>

```python
def age_group(a):
    if a < 30:
        return '18-29'
    elif a < 45:
        return '30-44'
    elif a < 60:
        return '45-59'
    else:
        return '60+'

df['age_group'] = df['age'].apply(age_group)
df['age_group'].value_counts()
```

#### <font color="darkgreen">Explanation</font>

- Used to create a new column called age_group.
- Groups customers into different age categories.
- Makes age-based analysis easier.

## Correlation of Dataset Features with Online Shopping Preference

```python
# Correlation bar chart showing top factors influencing online preference


# Compute correlations with Online Preference
df_encoded = pd.concat(
    [df.select_dtypes(include=['number']),
        pd.get_dummies(df['shopping_preference'], prefix='pref'),],
    axis=1,)
corr = (
    df_encoded.corr()['pref_Online']
    .drop(['pref_Store', 'pref_Online', 'pref_Hybrid'])
    .sort_values())

plt.figure(figsize=(9, 6))

plt.barh(corr.index, corr.values, color=colors)
plt.title('Feature Correlations with Online Shopping Preference',)
plt.xlabel('Correlation Coefficient with Online Preference')
plt.ylabel('Dataset Features')
plt.tight_layout()
plt.show()
```

### Insight

- Tech-savvy score shows the strongest positive association with Online Shopping Preference among the analyzed numerical variables..
- Monthly online orders show the second-highest positive correlation, at approximately 0.08.
- Online payment trust score and daily internet hours both show positive correlations of approximately 0.05.
- Average store spending has the strongest negative correlation with Online Shopping Preference, at approximately -0.49.
- Need-to-touch-and-feel score has a negative correlation of approximately -0.14.
- Monthly store visits also show a negative correlation of approximately -0.07.
- Most other features have correlations close to zero, indicating a very weak relationship with Online Shopping Preference.
- Overall, online-oriented behaviours such as tech-savviness, online orders, payment trust and internet usage are positively associated with online preference, while store spending and the need to touch and feel products are negatively associated.
